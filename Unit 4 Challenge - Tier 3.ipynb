**2.2. Cleaning the data**

You might find you need to transpose your dataframe, check out what its row indexes are, and reset the index. You  also might find you need to assign the values of the first row to your column headings  . (Hint: recall the .columns feature of DataFrames, as well as the iloc[] method).

```python
print(properties.columns)
properties.transpose()
```

```python
#First I'll fix the postcodes by adding them to a multi-index
prop_fix_postcodes1 = properties.transpose().set_index(properties.transpose().iloc[:,0], append=True)
#Now we drop the column and transpose again to rows, because online it said it is best to keep dates in rows
prop_fix_postcodes2 = prop_fix_postcodes1.drop(columns=prop_fix_postcodes1.columns[0]).transpose()
```

```python
prop_fix_dates = prop_fix_postcodes2.set_index(prop_fix_postcodes2.iloc[:,0], drop=True)
#prop_fix_dates has the dates as the new index
#Next we drop that old column
prop_new_indices = prop_fix_dates.drop(columns=prop_fix_dates.columns[0])
prop_new_indices
```

**2.3. Cleaning the data (part 2)**

You might we have to **rename** a couple columns. How do you do this? The clue's pretty bold...

```python
#Renaming the index as 'date'
temp1 = prop_new_indices.rename_axis('date')
#Renaming the columns as borough and postcode
temp2 = temp1.rename_axis(['borough','postcode'],axis="columns")
#Rename the boroughs to all uppercase so we don't have to be concerned about capitalization
prop_labeled = temp2.rename(str.upper, axis="columns", level=0)
```
