### Q7: Produce a list of all members who have used a tennis court. Include in your output the name of the court, and the name of the member formatted as a single column. Ensure no duplicate data, and order by the member name.
```
SELECT DISTINCT f.name AS courtname, CONCAT(m.firstname, ' ', m.surname) AS membername 
FROM bookings AS b 
INNER JOIN members AS m ON b.memid = m.memid 
INNER JOIN facilities AS f ON b.facid = f.facid 
WHERE f.facid IN (0,1) 
AND m.memid != 0 
ORDER BY m.surname,m.firstname;
```

````python
#I chose to order by member last name then member first name
cursor.execute("SELECT DISTINCT f.name AS courtname, CONCAT(m.firstname, ' ', m.surname) AS membername FROM bookings AS b INNER JOIN members AS m ON b.memid = m.memid INNER JOIN facilities AS f ON b.facid = f.facid WHERE f.facid IN (0,1) AND m.memid != 0 ORDER BY m.surname,m.firstname;")
nameandcourt = cursor.fetchall()
for item in nameandcourt:
    print(item[0] + ', ' + item[1])
````

### Q8: Produce a list of bookings on the day of 2012-09-14 which will cost the member (or guest) more than $30. Remember that guests have different costs to members (the listed costs are per half-hour 'slot'), and the guest user's ID is always 0. Include in your output the name of the facility, the name of the member formatted as a single column, and the cost. Order by descending cost, and do not use any subqueries.
```
SELECT f.name, CONCAT(m.firstname, ' ', m.surname) AS membername, 
    CASE WHEN b.memid = 0 THEN f.guestcost * b.slots 
    ELSE f.membercost * b.slots 
    END AS cost 
FROM bookings AS b 
INNER JOIN facilities AS f ON b.facid = f.facid  
INNER JOIN members AS m ON b.memid = m.memid 
WHERE b.starttime LIKE '2012-09-14%' 
AND ( 
    (m.memid = 0 AND b.slots * f.guestcost > 30) 
    OR (m.memid != 0 AND b.slots * f.membercost > 30) 
    ) 
ORDER BY cost DESC ;")
```

````python
cursor.execute("SELECT f.name, CONCAT(m.firstname, ' ', m.surname) AS membername, CASE WHEN b.memid = 0 THEN f.guestcost * b.slots ELSE f.membercost * b.slots END AS cost FROM bookings AS b INNER JOIN facilities AS f ON b.facid = f.facid INNER JOIN members AS m ON b.memid = m.memid WHERE b.starttime LIKE '2012-09-14%' AND ( (m.memid = 0 AND b.slots * f.guestcost > 30) OR (m.memid != 0 AND b.slots * f.membercost > 30) ) ORDER BY cost DESC ;")
daybookingsthirty = cursor.fetchall()
for booking in daybookingsthirty:
    print(booking)
````

### Q9: This time, produce the same result as in Q8, but using a subquery.
```
SELECT facility, membername, cost FROM (
    SELECT f.name AS facility, CONCAT(m.firstname, ' ', m.surname) AS membername,
        CASE WHEN b.memid = 0 THEN f.guestcost * b.slots 
        ELSE f.membercost * b.slots 
        END AS cost 
    FROM bookings AS b
    INNER JOIN facilities AS f ON b.facid = f.facid 
    INNER JOIN members AS m ON b.memid = m.memid 
    WHERE b.starttime LIKE '2012-09-14%'
) AS tablewithcosts
WHERE cost > 30 
ORDER BY cost DESC;
```

````python
cursor.execute("SELECT facility, membername, cost FROM (SELECT f.name AS facility, CONCAT(m.firstname, ' ', m.surname) AS membername, CASE WHEN b.memid = 0 THEN f.guestcost * b.slots ELSE f.membercost * b.slots END AS cost FROM bookings AS b INNER JOIN facilities AS f ON b.facid = f.facid INNER JOIN members AS m ON b.memid = m.memid WHERE b.starttime LIKE '2012-09-14%') AS tablewithcosts WHERE cost > 30 ORDER BY cost DESC;")
daybooksubq = cursor.fetchall()
for booking in daybooksubq:
    print(booking)
````

### Q10: Produce a list of facilities with a total revenue less than 1000. The output of facility name and total revenue, sorted by revenue. Remember that there's a different cost for guests and members!
```
SELECT facilityname, revenue 
FROM 
    (SELECT f.name AS facilityname, 
        CASE WHEN b.memid = 0 THEN SUM(f.guestcost * b.slots) 
        ELSE SUM(f.membercost * b.slots) 
        END AS revenue 
    FROM bookings AS b 
    INNER JOIN facilities AS f ON b.facid = f.facid 
    GROUP BY f.facid ) AS summedtable 
WHERE revenue < 1000 
ORDER BY revenue;")

```

````python
cursor.execute("SELECT facilityname, revenue FROM (SELECT f.name AS facilityname, CASE WHEN b.memid = 0 THEN SUM(f.guestcost * b.slots) ELSE SUM(f.membercost * b.slots) END AS revenue FROM bookings AS b INNER JOIN facilities AS f ON b.facid = f.facid GROUP BY f.facid ) AS summedtable WHERE revenue < 1000 ORDER BY revenue;")
print(cursor.fetchall())
````

### Q13: Find the facilities usage by month, but not guests
```
SELECT f.name, strftime('%m',b.starttime) AS month, SUM(b.slots) 
FROM bookings AS b 
JOIN facilities AS f ON b.facid = f.facid 
WHERE b.memid !=0 
GROUP BY month, b.facid;")

```

````python
cursor.execute("SELECT f.name, strftime('%m',b.starttime) AS month, SUM(b.slots) FROM bookings AS b JOIN facilities AS f ON b.facid = f.facid WHERE b.memid !=0 GROUP BY month, b.facid;")
facbymonth = cursor.fetchall()
for entry in facbymonth:
    print(f'Facility: {entry[0]}, Month: {entry[1]}, Slots: {entry[2]}')
````
